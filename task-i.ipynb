```python
import pennylane as qml
from pennylane import numpy as np
import matplotlib.pyplot as plt
```

```python
# Creating a quantum device with 5 qubits
dev = qml.device('default.qubit', wires=5)

@qml.qnode(dev)
def circuit():

    # a) 5 qubits are already initialized in |0> state by default
    # b) Applying Hadamard gate to every qubit
    for i in range(5):
        qml.Hadamard(wires=i)

    # c) Applying CNOT operations on specified qubit pairs
    for i in range(4):
        qml.CNOT(wires=[i,i+1])

    # d) Swapping qubits 0 and 4
    qml.SWAP(wires=[0,4])

    # e) Rotating X with pi/2 on qubit 2
    qml.RX(np.pi/2, wires=0)

    # Returning the state for visualization
    return qml.state()
print(qml.draw(circuit)())
```

# 2) Implement a second circuit with a framework of your choice:

    a. Apply a Hadmard gate to the first qubit
    
    b. Rotate the second qubit by pi/3 around X
    
    c. Apply Hadamard gate to the third and fourth qubit
    
    d. Perform a swap test between the states of the first and second qubit |q1 q2> and the third and fourth qubit |q3 q4>


```python
# Creating a quantum device with 5 qubits
dev = qml.device('default.qubit', wires=5)

@qml.qnode(dev)
def circuit_2():
    # a) Applying Hadamard gate to the first qubit
    qml.Hadamard(wires=0)
    
    # b) Rotating the second qubit by pi/3 around X
    qml.RX(np.pi/3, wires=1)
    
    # c) Applying Hadamard gate to the third and fourth qubits
    qml.Hadamard(wires=2)
    qml.Hadamard(wires=3)
    
    # d) Performing swap test
    # Using the fifth qubit as an ancilla qubit
    qml.Hadamard(wires=4)  # Prepare ancilla
    
    # Controlled swap operations
    qml.CSWAP(wires=[4, 0, 2])  # Swaping first pair
    qml.CSWAP(wires=[4, 1, 3])  # Swaping second pair
    
    qml.Hadamard(wires=4)  # Final Hadamard on ancilla
    
    # Measuring the ancilla qubit
    return qml.probs(wires=4)

# Execute the circuit
result = circuit_2()

print(qml.draw(circuit_2)())

# Print the probabilities of the ancilla qubit
print("Ancilla Qubit Probabilities:")
print(result)
```
