In a workshop 7-8 months back that I and my club mates were organizing for the first years of our college (on "Intermediate to NLP & Large Language Models"), we presented them summaries of SOTA architectures starting from Transformers to BERT and GPT all the way to Llama-family. One curious freshie asked a really good question: "Latest developments in LLMs all seem to be about scaling up existing models and optimizing the use of resources. Does that mean that we are still quite far from training models to their capacity? And if so, why can't we just scale them on quantum computers?"
It was a good question because it touched on two basic issues—the boundaries of classical computation and the state of quantum computing today. I remember answering something like: "Scaling works because deep learning isn't just about clever architectures, but it also envolves some degree of brute-force optimization on huge datasets. We aren't inefficiently training models—we're just hitting classical computing at its absolute limits. The fact that we haven't leaped straight to quantum is because, in spite of its theoretical benefits, practical quantum hardware is far from prepared to do deep learning at scale. Currently, we're still working with error rates, decoherence, and qubit scalability. If you think GPT-4 training costs are pricey, try running it on a quantum computer that can hardly muster 100 stable qubits. That's like attempting to train an LLM on an abacus."

While I was able to give a satisfactory answer to that curious student, I just couldn't convince myself about my perspective on quantum machine learning, it's fine that quantum computers are still so much error prone and expensive to produce but that shouldn't be the argument to justify why quantum machine learning models aren't actually competitive at this stage. There's always a starting point, like "how a simple idea of propagating a quantity backwards in a network and efficiently compute them for other parts of the network using the relationship between them" (Yes, Backpropagation) give rise to this whole world of Deep Learning. 
I shared my thoughts with a senior of our college who happened to have interned as Quantum Researcher at a reputated organization, he then gave me a new perspective: "The real motive isn’t just about running classical deep learning models on quantum hardwares but designing learning methods that truly takes advantage of quantum principles. When that happens, we might not just be scaling AI—we might be redefining what intelligence even looks like." That was something which has stayed with me till now. That event really made me rethink my research interests.
